## Creating a Tensor

```python
# using empty
```

- allocates memory, creates a space of the dimension mentioned in the memory.Doesnt assigns any value,since the value already exisits in that space ,it just shows it
- Creates an uninitialized tensor. It allocates memory but does not set values (contains random/garbage data already in memory).

```python
torch.empty(2,3) #2d tensor
```

```python
a =  torch.empty(3,3)
```

```python
type(a) # Check type
```

```python
# using zeros 
```

Creates a tensor filled entirely with zeros.Usually biases value of neural networks can be stored as zeroes inititally.

```python
torch.zeros(2,3)
```

```python
# using ones
```

Creates a tensor filled entirely with ones.

```python
torch.ones(2,3)
```

```python
# using rand
```

torch.rand(rows, columns): Creates a tensor filled with random numbers from a uniform distribution between 0 and 1.weight intitialization matrix can be initialized with this random numbers.

```python
torch.rand(2,3)
```

```python
# use of seed
```

Sets a fixed seed for generating random numbers. This ensures reproducibility (running the same seed guarantees the exact same random numbers every time).

```python
torch.manual_seed(100)
torch.rand(2,2) # same tensor reproduced again
```

```python
# using tensor
```

torch.tensor(data): Creates a tensor directly from standard Python data structures like lists or nested lists.Creates custom tensor.

```python
torch.tensor([[1,2,3],[4,5,6]])
```

```python
# arange
```

torch.arange(start, end, step): Creates a 1D tensor with a sequence of integers from start up to (but not including) end.

```python
torch.arange(0,10)
```

```python
# linspace
```

torch.linspace(start, end, steps): Creates a 1D tensor of evenly (linearly) spaced points between start and end (inclusive). The steps parameter defines exactly how many numbers to generate.

```python
torch.linspace(0,10,3)
```

```python
# eye
```

torch.eye(n): Creates an n x n 2D square identity matrix (contains 1. on the main diagonal and 0. everywhere else).

```python
torch.eye(8)
```

```python
# using full
```

torch.full(size, fill_value): Creates a tensor of the specified size (shape) filled entirely with your chosen fill_value.

```python
torch.full((3,3),9) # creates a 3x3 matrix where every single element is 9
```

```python
torch.full((6,3),3)
```

## Tensor Shapes

```python
x = torch.tensor([[2,3,4],[4,5,6]])
x
```

```python
x.shape # Returns the dimensions (size) of a tensor.
```

The _like suffix: Functions ending in _like automatically create a new tensor with the exact same shape and data type as an existing reference tensor.

```python
torch.empty_like(x) # Creates an uninitialized tensor matching the shape of the reference.
```

```python
torch.zeros_like(x) # Creates a tensor filled entirely with zeros matching the shape of the reference.
```

```python
torch.ones_like(x) # Creates a tensor filled entirely with ones matching the shape of the reference.
```

```python
torch.rand_like(x)
```

### Limitation: Generates a random tensor matching the reference shape, but it requires the reference tensor to be a floating-point data type.
