### Views vs copies
Many ops return views (no new memory), others create copies.
- Quick checks:
- y = x.view(-1)
- y.storage().data_ptr() == x.storage().data_ptr()  # often True for views


- Rules of thumb : view, transpose, permute → usually views (metadata changes)
clone, many advanced indexing patterns → copy

- Why it matters:
Views are memory-efficient, but can surprise you if you modify in-place.

```python
size = 10000
```

```python
# createrandom matrices on cpu
mat_cpu1 = torch.randn(size,size)
mat_cpu2 = torch.randn(size,size)
```

```python
# Matrix mulplication time on cpu
st_time = time.time()
matmul_cpu = torch.matmul(mat_cpu1,mat_cpu2)
cpu_time = time.time() - st_time
```

```python
cpu_time
```

```python
# move matrices to gpu
mat_gpu1 = mat_cpu1.to('cuda')
mat_gpu2 = mat_cpu2.to('cuda')
```

```python
# matrix multiplication time on gpu
st_time = time.time()
matmul_gpu =  torch.matmul(mat_gpu1,mat_gpu2)
gpu_time = time.time() - st_time
```
